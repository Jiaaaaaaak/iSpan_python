```python
# f-string (又稱作 formatted string literals, version >= 3.6)
name = 'Andy'
age = 18

msg = (f"{name} is {age} years old.")
print(msg)

print(f"{name} is {age} years old.")
```

```python
log = f"2025-11-01 15:52 [INFO] - {msg}" 
print(log)
```

```python
# 靠左對齊、靠右對齊

name = 'Andy'
age = 18
msg = f'[name: {name:X<10}], age:[{age:X>10}]'

print(msg)
```
