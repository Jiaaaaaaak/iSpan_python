```python
# 初始化一個 list
ids = [1, 2, 3, 4, 5, 6]
print(ids)
```

```python
# 新增元素在 list 尾端
# list.append()
ids = [1, 2, 3, 4, 5, 6]
print(ids)
ids.append(7)
print(ids)
```

```python
ids = [1, 2, 3, 4, 5, 6]

print(ids[3])
```

```python
# 修改索引位置的元素
ids = [1, 2, 3, 4, 5, 6]
print(ids)
ids[3] = 99
print(ids)
```

```python
# 刪除索引位置的元素
# ids.pop(<index索引位置>)
# 變數不用<>
ids = [1, 2, 3, 4, 5, 6]
ids.pop(3)
print(ids)
```

```python
# 在指定索引 1新增元素 9 ，其餘元素往後移
# list.insert(<index>, <value>)
ids = [1, 2, 3, 4, 5, 6]
ids.insert(1, 9)
print(ids)
```

```python
# 移除指定的值
# list.remove(<value>)
# 跟list.pop()不同，remove是直接輸入value
ids = [1, 2, 3, 4, 5, 6]
ids.remove(4)
print(ids)
```

```python
# 迴圈執行新增9
ids = [1, 2, 3, 4, 5, 6]
for i in range(5):
    ids.insert(1, 9)
print(ids)
```

```python
# 寫while迴圈重複remove直到沒有9

while 9 in ids:
    ids.remove(9)
print(ids)
```

```python
ids = [1, 2, 3, 4, 5, 6]
for i in range(5):
    ids.append(10)
print(ids)
# 計算 list 內的某個值 共有幾個
# list.count(<value>)
# 寫一個變數去接計算出來的值
cnt_10 = ids.count(10)
print(cnt_10)
```

```python
outside = []
inside = []

for i in range(1, 11):
    inside.append(i)
    # print(inside)
    outside.append(inside.copy())

    
print(outside)
```

```python
outside = []
for i in range(1, 11):
    inside = []
    # print(f"i: {i}")
    for c in range(1, i+1):
        # print(f"c: {c}")
        inside.append(c)
    outside.append(inside)

    
print(outside)
```

```python
# enumerate() 同時取得 index 與 value
ids = [1, 22, 38, 54, 65]
for index, value in enumerate(ids):
    print(f"index: {index}, value: {value}")
```

### List comprehension 列表推導式
`new_list = [expression for item in iterable [if condition]]`  
  
`list的名稱 = [表達式 for 元素 in 迭代對象 [if 判斷式(...的條件)]]`

```python
# list comprehension
new_list_1_to_10 = [i for i in range(1, 11)]
print(new_list_1_to_10)
```

```python
# 1~10 的二次方
data = [i**2  for i in range(1, 11)]
print(data)
```

```python
outside = []
for i in range(1, 11):
    inside = [c for c in range(1 ,i+1)]
    outside.append(inside)
print(outside)
```

```python
# 建立一個 1~50 的 list 只要偶數

num_list = []
for i in range(1, 51):
    if i %2 == 0:
        num_list.append(i)
print(num_list)
```

```python
num_list = [i for i in range(1, 51)if i % 2 == 0]
print(num_list)
```
