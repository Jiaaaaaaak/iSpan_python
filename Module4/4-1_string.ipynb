|string方法| 功能 |
|:---:|:---:|
|`str.replace()` | 替換字串 |
| `str.strip()`| 去除兩側空格 |
| `str.upper()`| 字串小寫變大寫 |
| `str.lower()`| 字串大寫變小寫 |

```python
# 字串變數初始化

```

```python
# 替換字串
# 方法 也是function的一種
# function不一定是方法
# 定義在class底下的function可以稱為method
'''
str.replace()
將 string 中的 str1 替換成 str2
'''
name = 'Alex'
new_name = name.replace('ex', 'len')
print(new_name)
print(name)

```

```python
# 去除兩側空格
'''
str.strip()
去除字串 string 左、右兩邊的空格
str.lstrip()左
str.rstrip()右
'''
data = '  |__data__|  '
print(data)

new_data = data.strip()
print(new_data)
```

```python
# 字串小寫變大寫
'''
str.lower()
將字串string的字母全部改成小寫
將字串 string 裡的字母全部改成小寫
'''
data = 'CAR'
new_data = data.lower()
print(new_data)
'''
str.upper()
將字串string的字母全部改成大寫
將字串 string 裡的字母全部改成大寫
'''
data2 = 'friend'
new_data2 = data2.upper()
print(new_data2)
```

```python
# 字串分割成 list
# str.split()
greeting = "hello world! hello John"
# str_list = greeting.split('!')
str_list = greeting.split()
print(str_list)
```
