```python
# 初始化 dict
fruit_price_dict = {
        "apple": 100, 
        "orange": 50, 
        "tomato": 30
    }

print(fruit_price_dict)
```

```python
# 印出 蘋果的價格
fruit_price_dict = {"apple": 100, "orange": 50, "tomato": 30}
print(fruit_price_dict["apple"])
```

```python
# 修改 橘子的價格
fruit_price_dict = {"apple": 100, "orange": 50, "tomato": 30}
fruit_price_dict["orange"] = 70
print(fruit_price_dict)
```

```python
# 刪除 番茄 del
fruit_price_dict = {
        "apple": 100, 
        "orange": 50, 
        "tomato": 30
    }
del fruit_price_dict["tomato"]
print(fruit_price_dict)
```

```python
# 刪除 番茄 dict.pop()會報錯
fruit_price_dict = {"apple": 100, "orange": 50, "tomato": 30}
fruit_price_dict.pop("tomato")
print(fruit_price_dict)
```

```python
# 建立一個 value 含有 dict 的 dict
user_info = {
    "name": "andy",
    "age": 20,
    "info": {
        "nickname":"an",
        "favorite_role":"Woody",
        "phone":[
            "0912345678",
            "0912356789"
        ]
    }
}
```

```python
# 取得 nickname
print(user_info["info"])
print(user_info["info"]["nickname"])
```

```python
# 取得所有 phone_number
print(user_info["info"]['phone'])
for phone_number in user_info["info"]["phone"]:
    print(phone_number)
```

```python
'''
改寫 fruit_dict 每個水果需要 price, quantity/amount, color
apple 有 綠色 和 紅色
orange 有 橘色
tomato 有 紅色
'''

fruit_dict = {
    "apple": {
        "price": 30,
        "amount":100,
        "color": ["green","red"]
    },
    "orange": {
        "price": 50,
        "amount": 30,
        "color": ["orange"]
    },
    "tomato": {
        "price": 80,
        "amount": 20,
        "color": ["red"]
    }
}
# 格式保持一致，顏色只有一種也用list

print(fruit_dict)
```

##### dict 的合併

```python
# dict 的合併
# dict.update()
fruit_dict2 = {
    "kiwi": {
        "price": 30,
        "amount": 20,
        "color": ["green"]
    }
}

fruit_dict.update(fruit_dict2)
print(fruit_dict)
```

```python
# 補充：
# 不使用舊的update，用**拆解dict，重新建一個dict，實務會直接覆蓋(用同一個變數名fruit_dict)
print(fruit_dict)
fruit_dict3 = {
    **fruit_dict,
    "kiwi": {
        "price": 30,
        "amount": 20,
        "color": ["yellow"]
    },
}
print(fruit_dict3)
```

##### 另外的 print 方式 `pprint` (pretty-print) 

```python
# 從 Module 引入 function
from pprint import pprint
pprint(fruit_dict3)

# 打印出來的資料沒有按照順序
# 非基本語法，一定要引入套件
```

```python
'''
新增顏色
apple 有新的品種 顏色要有黃色
tomato 有綠色
'''
fruit_dict3['apple']['color'].append('yellow')
fruit_dict3['tomato']['color'].append('green')

pprint(fruit_dict3)

```
