### lambda 匿名函式  
用法  

`function_name = lanbda arguments: expression`

````python
# 建立函式
def add_num(x, y):
    return x + y

print(add_num(1, 5))
````

````python
# 基本用法 實務不會用在簡短之處
add_num = lambda x, y: x + y
print(add_num(1, 5))
````

````python
# 條件判斷
def check_even(num):
    if num % 2 == 0:
        return True
    else:
        return False

print(check_even(5))
````

````python
check_even = lambda num: True if num % 2 == 0 else False
print(check_even(5))
````

請建立一個 list ，並在該 list 內產生多個 `[1, 2, x]` 的 list (x 的值要遞增)  
x 到 10 即可  
Ex :  
```Python
[ [1, 2, 1], [1, 2, 2], [1, 2, 3]...[1, 2, 10] ]
```

````python
outside = []
for i in range(1, 11):
    outside.append([1, 2,i])


print(outside)

````

##### `list.sort()` 方法

````python
x = [2, 5, 1, 6, 10, 100]
x.sort() #預設從小排到大
print(x)
x.sort(reverse=True) #reverse=True從大排到小
print(x)
print(x[-3:])
````
